```python
from scipy import stats
from statsmodels.stats.power import TTestIndPower
import numpy as np
# H0: процентная ставка int_rate у дефолтных loan_status <= у выплаченных
# H1: процентная ставка int_rate у дефолтных loan_status > у выплаченных
print('H0: процентная ставка int_rate у дефолтных loan_status <= у выплаченных')
print('H1: процентная ставка int_rate у дефолтных loan_status > у выплаченных')

paid_statuses = ['Fully Paid', 'Does not meet the credit policy. Status:Fully Paid']
default_statuses = ['Charged Off', 'Default', 'Late (31-120 days)','Late (16-30 days)', 'Does not meet the credit policy. Status:Charged Off']

paid = df_nnn[df_nnn['loan_status'].isin(paid_statuses)]['int_rate'].dropna()
defaulted = df_nnn[df_nnn['loan_status'].isin(default_statuses)]['int_rate'].dropna()

alpha_levels = [0.1, 0.05, 0.01]
for alpha in alpha_levels:
    t_stat, p_val = stats.ttest_ind(defaulted, paid, alternative='greater', equal_var=True)
    print('alpha ', alpha)
    print('p-value ', f"{p_val:.6f}")
    if p_val < alpha:
        print('Отвергаем Н0')
    else:
        print('Не отвергаем Н0')
print('T-статистика', t_stat)
mean_paid = paid.mean()
mean_default = defaulted.mean()
print(f'Средняя ставка выплаченных: {mean_paid}')
print(f'Средняя ставка дефолтных: {mean_default}')
std_pooled = np.sqrt(((len(paid)-1) * np.var(paid) + (len(defaulted)-1) * np.var(defaulted)) / (len(paid)+len(defaulted)-2)) # общее стандартное отклонение
effect_size = (mean_default - mean_paid) / std_pooled

analysis = TTestIndPower()
for alpha in alpha_levels:
    power = analysis.solve_power(effect_size=effect_size, nobs1=len(defaulted), alpha=alpha, ratio=len(paid)/len(defaulted))
    print(f'Мощность критерия при α={alpha}: {power}')
print(f'Размер выборки выплаченных займов {len(paid)}')
print(f'Размер выборки дефолтных займов {len(defaulted)}')
print(f'Общий размер выборки {len(paid) + len(defaulted)}')
```

```python
# H0: Среднее количество просрочек у заемщиков с рейтингом A =< с рейтингом B
# H1: Среднее количество просрочек у заемщиков с рейтингом A > с рейтингом B

print('H0: Среднее количество просрочек у заемщиков с рейтингом A =< с рейтингом B')
print('H1: Среднее количество просрочек у заемщиков с рейтингом A > с рейтингом B')

grade_a_delinq = df_nnn[df_nnn['grade'] == 'A']['delinq_2yrs'].dropna()
grade_b_delinq = df_nnn[df_nnn['grade'] == 'B']['delinq_2yrs'].dropna()

alpha_levels = [0.1, 0.05, 0.01]
for alpha in alpha_levels:
    t_stat, p_val = stats.ttest_ind(grade_a_delinq, grade_b_delinq, alternative='greater', equal_var=True)
    print('alpha ', alpha)
    print('p-value ', f"{p_val:.6f}")
    if p_val < alpha:
        print('Отвергаем Н0')
    else:
        print('Не отвергаем Н0')
print('T-статистика', t_stat)
mean_b = grade_b_delinq.mean()
mean_a = grade_a_delinq.mean()
print(f'Среднее количество просрочек b: {mean_b}')
print(f'Среднее количество просрочек a: {mean_default}')
std_pooled = np.sqrt(((len(grade_b_delinq)-1) * np.var(grade_b_delinq) + (len(grade_a_delinq)-1) * np.var(grade_a_delinq)) / (len(grade_b_delinq)+len(grade_a_delinq)-2))
effect_size = (mean_a - mean_b) / std_pooled

analysis = TTestIndPower()
for alpha in alpha_levels:
    power = analysis.solve_power(effect_size=effect_size, nobs1=len(grade_a_delinq), alpha=alpha)
    print(f'Мощность критерия при alpha={alpha}: {power}')
print(f'Размер выборки просрочек b {len(grade_b_delinq)}')
print(f'Размер выборки просрочек a {len(grade_a_delinq)}')
print(f'Общий размер выборки {len(grade_b_delinq) + len(grade_a_delinq)}')

```

```python
# H0: Годовой доход собственников = годовой доход арендаторов  
# H1: Годовой доход собственников != годовой доход арендаторов

print('H0: Годовой доход собственников = годовой доход арендаторов')
print('H1: Годовой доход собственников != годовой доход арендаторов')

own_inc = df_nnn[df_nnn['home_ownership'] == 'OWN']['annual_inc'].dropna()
rent_inc = df_nnn[df_nnn['home_ownership'] == 'RENT']['annual_inc'].dropna()


alpha_levels = [0.1, 0.05, 0.01]
for alpha in alpha_levels:
    t_stat, p_val = stats.ttest_ind(own_inc, rent_inc, alternative='two-sided', equal_var=True)
    print('alpha ', alpha)
    print('p-value ', f"{p_val:.6f}")
    if p_val < alpha:
        print('Отвергаем Н0')
    else:
        print('Не отвергаем Н0')
print('T-статистика', t_stat)
mean_own = own_inc.mean()
mean_rent = rent_inc.mean() 
print(f'Средний годовой доход для own: {mean_own}')
print(f'Средний годовой доход для rent: {mean_rent}')

std_pooled = np.sqrt(((len(own_inc)-1) * np.var(own_inc, ddof=1) + (len(rent_inc)-1) * np.var(rent_inc, ddof=1)) / (len(own_inc) + len(rent_inc) - 2))
effect_size = (mean_own - mean_rent) / std_pooled
analysis = TTestIndPower()
for alpha in alpha_levels:
    power = analysis.solve_power(effect_size=abs(effect_size), nobs1=len(own_inc), alpha=alpha, ratio=len(rent_inc)/len(own_inc))
    print(f'Мощность критерия при alpha={alpha}: {power}')
    
print(f'Размер выборки собственников: {len(own_inc)}')
print(f'Размер выборки арендаторов: {len(rent_inc)}')
print(f'Общий размер выборки: {len(rent_inc)+len(own_inc)}')

```

```python
paid_statuses = ['Fully Paid', 'Does not meet the credit policy. Status:Fully Paid']
default_statuses = ['Charged Off', 'Default', 'Late (31-120 days)', 'Late (16-30 days)','Does not meet the credit policy. Status:Charged Off']

paid = df_nnn[df_nnn['loan_status'].isin(paid_statuses)]['int_rate'].dropna()
defaulted = df_nnn[df_nnn['loan_status'].isin(default_statuses)]['int_rate'].dropna()

alpha_levels = [0.1, 0.05, 0.01]
print('----------------------------')
for alpha in alpha_levels:
    print('alpha ', alpha)
    print('----------------------------')
    # Критерий Фишера
    f_stat = np.var(defaulted, ddof=1) / np.var(paid, ddof=1)
    dfn = len(defaulted) - 1
    dfd = len(paid) - 1
    p_val_fisher = 2 * min(stats.f.cdf(f_stat, dfn, dfd), 1 - stats.f.cdf(f_stat, dfn, dfd))

    print('Критерий Фишера')
    print('F-статистика ', f_stat)
    print('p-value',f'{p_val_fisher:.6f}')
    if p_val_fisher < alpha:
        print('Отвергаем H0')
    else:
        print('Не отвергаем H0')
    print('----------------------------')


    # Критерий Левене
    w_stat, p_val_levene = stats.levene(defaulted, paid, center='median')
    print('Критерий Левене')
    print('W-статистика ', w_stat)
    print('p_value', f'{p_val_levene:.6f}')
    if p_val_levene < alpha:
        print('Отвергаем H0')
    else:
        print('Не отвергаем H0')
    print('----------------------------')
    # Критерий Бартлетта
    b_stat, p_val_bartlett = stats.bartlett(defaulted, paid)
    print('Критерий Бартлетта')
    print('B-статистика ', b_stat)
    print('p_value', f'{p_val_bartlett:.6f}')
    if p_val_bartlett < alpha:
        print('Отвергаем H0')
    else:
        print('Не отвергаем H0')
    print('----------------------------')

    # Критерий Флигнера–Килина
    fk_stat, p_val_fligner = stats.fligner(defaulted, paid)
    print('Критерий Флигнера–Килина')
    print('FK-статистика ', fk_stat)
    print('p_value', f'{p_val_fligner:.6f}')
    if p_val_fligner < alpha:
        print('Отвергаем H0')
    else:
        print('Не отвергаем H0')
    print('----------------------------')
```

а) Критерий Стьюдента, применим к нормально распределенным данным

    1. int_rate
    H0: процентная ставка int_rate у дефолтных loan_status <= у выплаченных
    H1: процентная ставка int_rate у дефолтных loan_status > у выплаченных
    Отвергаем H0, следовательно, дефолтные заемщики действительно имеют более высокие ставки, t-тест - односторонний, мощность = 1, выборка очень большая и результат надежный

    2. delinq_2yrs
    H0: Среднее количество просрочек у заемщиков с рейтингом A <= с рейтингом B
    H1: Среднее количество просрочек у заемщиков с рейтингом A > с рейтингом B, t-тест - односторонний, t-статистика отрицательная, следовательно, просрочек у А меньше, чем у В          
    мощность = 1 - результат надежный, не отвеграем H0 - значит у рейтинга А меньше просрочек, чем у рейтинга В

    3. annual_inc
    H0: Годовой доход собственников = годовой доход арендаторов  
    H1: Годовой доход собственников != годовой доход арендаторов
    t-тест - двухсторонний, мощность = 1 - результат надежный, Отвергаем H0, следовательно, доход собственников выше, чем у арендаторов

b) Критерий Уилкоксона-Манна-Уитни, применим для ненормальных данных

    1. int_rate
    Во всех случаях отвергаем H0 - дефолтные кредиты имеют существенно более высокие процентные ставки, U-тест подтверждает результат t-теста.

    2. delinq_2yrs
    p_val = 1.0, следовательно, нет оснований отвергать H0, различий в количестве просрочек не выявлено и рейтинг A не хуже B, U-тест подтверждает результат t-теста.

    3. annual_inc
    Во всех случаях отвергаем H0, значит доходы собственников статистически выше. U-тест подтверждает результат t-теста.

Данные int_rate, delinq_2yrs, annual_inc не являются нормально распредлеленными, поэтому критерий Стьюдента к ним не применим, однако 
критерий Уилкоксона-Манна-Уитни уже подходит и его можем применить.

    с) Критерии Фишера, Левене, Бартлетта, Флигнера-Килина (проверка гипотез об однородности дисперсий)

    1. int_rate
    Во всех критериях (Фишера, Левене, Бартлетта, Флигнера–Килина) p-value < alpha при alpha = 0.1, 0.05 и 0.01, дисперсии существенно 
    различаются, то есть группы имеют разный разброс значений.

    2. delinq_2yrs 
    Аналогично, каждый из критериев отвергает гипотезу о равенстве дисперсий - наблюдается значительный разброс между выборками
        
    3. annual_inc
    Отвергаем гипотезу и о равенстве дисперсий - они различны, следовательно, неоднороджны по степени выборки

    Критерии Фишера и Бартлетта применимы только к нормлаьным данным, Левене и Флигнера-Килина применимы к данным, не только нормальным
    Поэтому к int_rate, delinq_2yrs, annual_inc применимы только критерии Левене и Флигнера-Килина. Причем критерий Флигнера-Килина имеет

2) Метод точного теста Фишера

    Применим только к таблицам вида 2х2, поэтому можем проверять связь только бинарных признаков, в моем датасете только 2 бинарных признака
    поэтому проверяем связь между двумя бинарными признаками - term и application_type (срок займа и тип заявки на кредит)
    итог - тип заявки на кредит влияет на срок кредита - Individual чаще берут кредит на 36 месяцев относительно 60 месяцев, 
    по сравнению с Joint App

```python
# Метод теста МакНемара
from statsmodels.stats.contingency_tables import mcnemar
df_mcn = df[['loan_status', 'grade']].dropna()
problem_status = ['Charged Off', 'Default', 'Late (16-30 days)', 'Late (31-120 days)']
df_mcn['loan_status_bin'] = df_mcn['loan_status'].isin(problem_status).astype(int)
df_mcn['high_grade'] = df_mcn['grade'].isin(['A','B','C']).astype(int)

tbl = pd.crosstab(df_mcn['loan_status_bin'], df_mcn['high_grade'])
print(tbl)
result = mcnemar(tbl, exact=True)  # exact=True, чтобы избежать проблем с малым количеством
print(f'Статистика = {result.statistic}')
print(f'p-value = {result.pvalue:.4g}')
```

3.  Метод теста МакНемара 

    Этот метот требует две бинарные переменные, представляющие парные измерения одного и того же признака на одном объекте, 
    поэтому loan_status и grade привела к бинарному виду, итог - между кредитным рейтингом и статусом кредита существует связь

```python
# Метод Кохрана-Мантеля-Хензеля
from statsmodels.stats.contingency_tables import StratifiedTable

df_cmh = df[['loan_status', 'grade', 'term']].dropna()
problem_status = ['Charged Off', 'Default', 'Late (16-30 days)', 'Late (31-120 days)']
df_cmh['loan_status_bin'] = df_cmh['loan_status'].isin(problem_status).astype(int)

tables = []
for t, group in df_cmh.groupby('term'):
    tbl = pd.crosstab(group['grade'].isin(['A','B','C']).astype(int), group['loan_status_bin'])
    tables.append(tbl.values)
    print(f'\nСрок кредита {t}:')
    print(tbl)

cmh = StratifiedTable(tables)
test_res = cmh.test_null_odds()

print(f'CMH статистика = {test_res.statistic:.4f}')
print(f'p-value = {test_res.pvalue:.4f}')
```

4. Метод Кохрана-Мантеля-Хензеля

    Этот метод проверяет связь между двумя бинарными признаками с учётом влияния третьего, поэтому так же loan_status и grade привела к бинарному виду
    итог - существует связь между рейтингом кредита и вероятностью проблемного статуса , даже после контроля по сроку кредита, то есть
    высокая оценка кредита означает что проблемный статус кредита наименее вероятнет независимо от срока кредита.

```python
# Линейная регрессия (линейная модель)
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score
import pandas as pd

df_clean = df_nnn.dropna()
y = df_clean['loan_amnt']
features = ['delinq_2yrs', 'int_rate', 'annual_inc', 'acc_open_past_24mths', 'installment', 'total_acc']
X = df_clean[features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print('Линейная регрессия')
print(f'MSE: {mse:.6f}')
print(f'R^2: {r2:.6f}')
```

```python
residuals = y_test - y_pred
plt.figure(figsize=(8,6))
sns.scatterplot(x=y_pred, y=residuals)
plt.axhline(0, color='r', linestyle='--')
plt.xlabel('Предсказанные значения')
plt.ylabel('Ошибки предсказания')
plt.title('Остатки vs предсказания (Линейная регрессия)')
plt.show()
```

```python
# Полиномиальная регрессия (нелинейная модель)
from sklearn.preprocessing import PolynomialFeatures

df_clean = df_nnn.dropna()
y = df_clean['loan_amnt']

features = ['delinq_2yrs', 'int_rate', 'annual_inc', 'acc_open_past_24mths', 'installment', 'total_acc']
X = df_clean[features]
poly = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X_poly, y, test_size=0.3, random_state=42)

model_poly = LinearRegression()
model_poly.fit(X_train, y_train)
y_pred_poly = model_poly.predict(X_test)

mse = mean_squared_error(y_test, y_pred_poly)
r2 = r2_score(y_test, y_pred_poly)

print('Полиномиальная регрессия')
print(f'MSE: {mse:.6f}')
print(f'R^2: {r2:.6f}')
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(y_test, y_pred_poly, alpha=0.5, label='Предсказания')

min_val = min(y_test.min(), y_pred_poly.min())
max_val = max(y_test.max(), y_pred_poly.max())
plt.plot([min_val, max_val], [min_val, max_val], 'r--', linewidth=2, label='Идеальный прогноз')

plt.xlabel('Фактические значения')
plt.ylabel('Предсказанные значения')
plt.title(f'Факт vs Предсказание(Полиномиальная регрессия)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

Полиномиальная и линейная регрессия показали почти одинаковую высокую точность предсказаний, однако у полиномиальной регрессии результаты
